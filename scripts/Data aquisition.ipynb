```python
import os
import csv

raw_folder = "data/raw_transcripts"
metadata_file = "data/metadata.csv"

company_info = {
    "TCS":      ("TATA", "IT"),
    "HDFC":  ("HDFC Bank", "Banking"),
    "RELIANCE":  ("Reliance Industries", "Conglomerate"),
}

rows = []
for filename in os.listdir(raw_folder):
    if filename.endswith(".pdf"):
        parts = filename.replace(".pdf", "").split("_")
        ticker = parts[0]
        quarter = parts[1]

        company_name, sector = company_info.get(ticker, ("Unknown", "Unknown"))
        filepath = os.path.join(raw_folder, filename)
        filesize_kb = round(os.path.getsize(filepath) / 1024, 1)

        rows.append({
            "filename": filename,
            "ticker": ticker,
            "company": company_name,
            "sector": sector,
            "quarter": quarter,
            "filepath": filepath,
            "size_kb": filesize_kb
        })

with open(metadata_file, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=rows[0].keys())
    writer.writeheader()
    writer.writerows(rows)

print(f"✅ Metadata created: {len(rows)} transcripts logged")
```

```python
import os

raw_folder = "data/raw_transcripts"

rename_map = {
    "BAJAJ_Q1FY25.pdf": "BAJAJFIN_Q1FY25.pdf",
    "BAJAJ_Q2FY25.pdf": "BAJAJFIN_Q2FY25.pdf",
    "BAJAJ_Q3FY25.pdf": "BAJAJFIN_Q3FY25.pdf",
    "BAJAJ_Q4FY25.pdf": "BAJAJFIN_Q4FY25.pdf",
    "Infosys_Q1FY.pdf": "INFY_Q1FY25.pdf",
    "Infosys_Q2FY.pdf": "INFY_Q2FY25.pdf",
    "Infosys_Q3FY.pdf": "INFY_Q3FY25.pdf",
    "Infosys_Q4FY.pdf": "INFY_Q4FY25.pdf",
    "MarutiSuzuki_Q1FY25.pdf": "MARUTI_Q1FY25.pdf",
    "MarutiSuzuki_Q2FY25.pdf": "MARUTI_Q2FY25.pdf",
    "MarutiSuzuki_Q3FY25.pdf": "MARUTI_Q3FY25.pdf",
    "MarutiSuzuki_Q4FY25.pdf": "MARUTI_Q4FY25.pdf",
}

for old_name, new_name in rename_map.items():
    old_path = os.path.join(raw_folder, old_name)
    new_path = os.path.join(raw_folder, new_name)
    if os.path.exists(old_path):
        os.rename(old_path, new_path)
        print(f"✅ Renamed: {old_name} → {new_name}")
    else:
        print(f"⚠️ Not found: {old_name}")
```

```python
import os
import csv

raw_folder = "data/raw_transcripts"
metadata_file = "data/metadata.csv"

company_info = {
    "HDFC":      ("HDFC Bank", "Banking"),
    "HUL":       ("Hindustan Unilever", "FMCG"),
    "Tata":      ("Tata Group", "Conglomerate"),
    "RELIANCE":  ("Reliance Industries", "Conglomerate"),
    "WIPRO":     ("Wipro", "IT"),
    "INFY":      ("Infosys", "IT"),
    "TCS":       ("TCS", "IT"),
    "BAJAJFIN":  ("Bajaj Finance", "NBFC"),
    "MARUTI":    ("Maruti Suzuki", "Auto"),
    "SUNPHARMA": ("Sun Pharma", "Pharma"),
}

rows = []
for filename in sorted(os.listdir(raw_folder)):
    if not filename.endswith(".pdf"):
        continue
    parts = filename.replace(".pdf", "").split("_")
    if len(parts) < 2:
        print(f"⚠️ Skipping bad filename: {filename}")
        continue

    ticker = parts[0]
    quarter = parts[1]
    company_name, sector = company_info.get(ticker, ("Unknown", "Unknown"))
    filepath = os.path.join(raw_folder, filename)
    filesize_kb = round(os.path.getsize(filepath) / 1024, 1)

    rows.append({
        "filename": filename,
        "ticker": ticker,
        "company": company_name,
        "sector": sector,
        "quarter": quarter,
        "filepath": filepath,
        "size_kb": filesize_kb
    })

with open(metadata_file, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=rows[0].keys())
    writer.writeheader()
    writer.writerows(rows)

print(f"✅ Metadata created: {len(rows)} transcripts")
print("\nBy company:")
import pandas as pd
df = pd.read_csv(metadata_file)
print(df.groupby(['ticker','sector']).size())
```
