```python
import pdfplumber
import os
import json

raw_folder = "data/raw_transcripts"
processed_folder = "data/processed"

os.makedirs(processed_folder, exist_ok=True)

results = []

for filename in os.listdir(raw_folder):
    if not filename.endswith(".pdf"):
        continue
    
    filepath = os.path.join(raw_folder, filename)
    output_name = filename.replace(".pdf", ".txt")
    output_path = os.path.join(processed_folder, output_name)
    
    full_text = ""
    
    with pdfplumber.open(filepath) as pdf:
        total_pages = len(pdf.pages)
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                full_text += text + "\n"
    
    # Save as text file
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(full_text)
    
    results.append({
        "filename": filename,
        "pages": total_pages,
        "characters": len(full_text),
        "output": output_name
    })
    
    print(f"✅ {filename} → {total_pages} pages, {len(full_text):,} characters")

print(f"\n🎉 Done! {len(results)} files extracted and saved to data/processed/")
```

```python
import os

raw_folder = "data/raw_transcripts"
processed_folder = "data/processed"
os.makedirs(processed_folder, exist_ok=True)

# Clear old processed files first
for f in os.listdir(processed_folder):
    if f.endswith(".txt"):
        os.remove(os.path.join(processed_folder, f))
print("🗑️ Cleared old processed files")

results = []
for filename in sorted(os.listdir(raw_folder)):
    if not filename.endswith(".pdf"):
        continue

    filepath = os.path.join(raw_folder, filename)
    output_path = os.path.join(processed_folder, filename.replace(".pdf", ".txt"))

    full_text = ""
    with pdfplumber.open(filepath) as pdf:
        total_pages = len(pdf.pages)
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                full_text += text + "\n"

    with open(output_path, "w", encoding="utf-8") as f:
        f.write(full_text)

    results.append({
        "filename": filename,
        "pages": total_pages,
        "characters": len(full_text)
    })
    print(f"✅ {filename} → {total_pages} pages, {len(full_text):,} chars")

print(f"\n🎉 Done! {len(results)} files extracted")
```
